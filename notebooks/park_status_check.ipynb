4. แปลงช่วงเวลาปิดตามฤดูกาล

ปรับข้อมูลช่วงปิดให้อยู่ในรูปแบบวันและเดือน
เพื่อใช้ตรวจว่าแหล่งท่องเที่ยวปิดในช่วงใดของปี
โดยไม่ผูกข้อมูลกับปีใดปีหนึ่ง เนื่องจากเป็นช่วงปิดตามฤดูกาล

```python
import numpy as np
import re

thai_month_map = {
    "ม.ค.": 1,
    "ก.พ.": 2,
    "มี.ค.": 3,
    "เม.ย.": 4,
    "พ.ค.": 5,
    "มิ.ย.": 6,
    "ก.ค.": 7,
    "ส.ค.": 8,
    "ก.ย.": 9,
    "ต.ค.": 10,
    "พ.ย.": 11,
    "ธ.ค.": 12
}

def parse_thai_day_month(value):
    if pd.isna(value):
        return pd.Series([np.nan, np.nan])

    value = str(value).strip()

    # กรณีไม่มีวันที่ระบุ
    if value in ["-", ""]:
        return pd.Series([np.nan, np.nan])

    # เช่น 1-มี.ค.
    match = re.match(r"^(\d{1,2})-(.+)$", value)

    if not match:
        return pd.Series([np.nan, np.nan])

    day = int(match.group(1))
    month_text = match.group(2).strip()

    month = thai_month_map.get(month_text, np.nan)

    return pd.Series([day, month])


status[
    ["close_start_day", "close_start_month"]
] = status["วันที่เริ่มปิด"].apply(parse_thai_day_month)

status[
    ["close_end_day", "close_end_month"]
] = status["วันที่สิ้นสุดการปิด"].apply(parse_thai_day_month)
```

6. สร้างสถานะเปิด-ปิดรายเดือน

แปลงข้อมูลช่วงปิดตามฤดูกาลให้อยู่ในรูปแบบรายเดือน
โดยกำหนดสถานะเป็น Open, Closed หรือ Unknown
เพื่อเตรียมนำไปสรุปเป็นระดับอุทยาน

```python
def get_month_status(row, month):
    
    # เปิดตลอดปี
    if row["เปิดตลอดปี"] == "ใช่":
        return "Open"
    
    start_month = row["close_start_month"]
    end_month = row["close_end_month"]
    
    # ไม่เปิดตลอดปี แต่ไม่มีข้อมูลช่วงปิด
    if pd.isna(start_month) or pd.isna(end_month):
        return "Unknown"
    
    start_month = int(start_month)
    end_month = int(end_month)
    
    # ช่วงปิดอยู่ภายในปีเดียวกัน
    # เช่น มิ.ย. - ก.ย.
    if start_month <= end_month:
        if start_month <= month <= end_month:
            return "Closed"
        return "Open"
    
    # ช่วงปิดข้ามปี
    # เช่น พ.ย. - ก.พ.
    else:
        if month >= start_month or month <= end_month:
            return "Closed"
        return "Open"
```

```python
monthly_status_rows = []

for _, row in status.iterrows():
    
    for month in range(1, 13):
        
        monthly_status_rows.append({
            "park_name": row["ชื่อหน่วยงาน"],
            "attraction_name": row["ชื่อแหล่งท่องเที่ยว"],
            "province": row["จังหวัด"],
            "latitude": row["latitude"],
            "longitude": row["longitude"],
            "month": month,
            "status": get_month_status(row, month)
        })

attraction_monthly_status = pd.DataFrame(monthly_status_rows)

attraction_monthly_status.head(20)
```

7. สรุปสถานะเปิด-ปิดระดับอุทยาน

รวมสถานะของแหล่งท่องเที่ยวภายในแต่ละอุทยานเป็นรายเดือน
เพื่อดูจำนวนจุดที่เปิด ปิด หรือไม่มีข้อมูล และคำนวณสัดส่วนความพร้อม
สำหรับการท่องเที่ยวในแต่ละเดือน

```python
park_month_availability = (
    attraction_monthly_status
    .groupby(["park_name", "month"])
    .agg(
        total_attractions=("attraction_name", "count"),
        open_attractions=("status", lambda x: (x == "Open").sum()),
        closed_attractions=("status", lambda x: (x == "Closed").sum()),
        unknown_attractions=("status", lambda x: (x == "Unknown").sum())
    )
    .reset_index()
)

# จำนวนจุดที่เรารู้สถานะจริง
park_month_availability["known_attractions"] = (
    park_month_availability["open_attractions"] +
    park_month_availability["closed_attractions"]
)

# % จุดที่เปิด จากเฉพาะจุดที่ทราบสถานะ
park_month_availability["availability_pct"] = (
    park_month_availability["open_attractions"]
    / park_month_availability["known_attractions"]
    * 100
)

# % ข้อมูลที่ยังไม่ทราบสถานะ
park_month_availability["unknown_pct"] = (
    park_month_availability["unknown_attractions"]
    / park_month_availability["total_attractions"]
    * 100
)

park_month_availability.head(20)
```

10. ปรับรูปแบบชื่ออุทยานก่อนเชื่อมข้อมูลเปิด-ปิด

ปรับชื่ออุทยานจากข้อมูลนักท่องเที่ยวและข้อมูลเปิด-ปิด
ให้อยู่ในรูปแบบมาตรฐานเดียวกัน เพื่อลดปัญหาจากช่องว่าง
เครื่องหมายขีด คำนำหน้า และสถานะเตรียมการ

```python
import re

def normalize_park_name(name):
    if pd.isna(name):
        return None

    name = str(name).strip()

    # ทำ Unicode ให้เป็นรูปแบบเดียวกัน
    name = pd.Series([name]).str.normalize("NFC").iloc[0]

    # ลบคำว่า "อุทยานแห่งชาติ" หน้าชื่อ ถ้ามี
    name = re.sub(r"^อุทยานแห่งชาติ\s*", "", name)

    # ลบสถานะเตรียมการ
    name = re.sub(r"\s*\(เตรียมการฯ?\)", "", name)

    # ลบช่องว่างล่องหน
    name = name.replace("\u200b", "")

    # จัดเครื่องหมายขีดให้เหมือนกัน
    name = re.sub(r"\s*[-–—]\s*", "-", name)

    # ช่องว่างหลายตัวให้เหลือหนึ่ง
    name = re.sub(r"\s+", " ", name)

    return name.strip()


tourism["park_status_key"] = (
    tourism["park_name"]
    .apply(normalize_park_name)
)

status["park_status_key"] = (
    status["ชื่อหน่วยงาน"]
    .apply(normalize_park_name)
)


matched_after_clean = set(tourism["park_status_key"]) & set(status["park_status_key"])

unmatched_after_clean = sorted(
    set(tourism["park_status_key"]) -
    set(status["park_status_key"])
)

print("จำนวนอุทยานที่จับคู่ได้หลังปรับชื่อ:")
print(len(matched_after_clean))

print("\nจำนวนที่ยังจับคู่ไม่ได้:")
print(len(unmatched_after_clean))

print("\nรายชื่อ:")
print(unmatched_after_clean)
```

11. ตรวจชื่ออุทยานที่ยังจับคู่ไม่ได้

ค้นหาชื่อในข้อมูลเปิด-ปิดที่มีความใกล้เคียงกับอุทยานที่ยังจับคู่ไม่ได้
เพื่อใช้ตรวจสอบชื่อเก่า ชื่อใหม่ หรือรูปแบบการเรียกที่แตกต่างกัน
ก่อนกำหนด Mapping แบบเจาะจง

```python
from difflib import SequenceMatcher

status_park_names = (
    status["park_status_key"]
    .dropna()
    .unique()
)

match_results = []

for name in unmatched_after_clean:

    best_match = None
    best_score = 0

    for candidate in status_park_names:

        score = SequenceMatcher(
            None,
            name,
            candidate
        ).ratio()

        if score > best_score:
            best_score = score
            best_match = candidate

    match_results.append({
        "tourism_name": name,
        "closest_status_name": best_match,
        "similarity": round(best_score, 2)
    })

status_name_matches = pd.DataFrame(match_results)

status_name_matches.sort_values(
    "similarity",
    ascending=False
)
```

12. กำหนด Mapping สำหรับชื่ออุทยานที่ตรวจสอบแล้ว

ปรับชื่อเฉพาะกรณีที่สามารถระบุได้ว่าเป็นอุทยานเดียวกัน
โดยไม่ใช้ Fuzzy Matching รวมชื่อโดยอัตโนมัติ
เพื่อป้องกันการเชื่อมข้อมูลผิดระหว่างอุทยานคนละแห่ง

```python
# Mapping เฉพาะชื่อที่ตรวจสอบแล้ว
status_mapping = {
    "ถ้ำปลา-น้ำตกผาเสื่อ": "ถ้ำปลา-ผาเสื่อ",
    "ไม้กลายเป็นหิน": "ดอยสอยมาลัย-ไม้กลายเป็นหิน",
    "ภูสวนทราย (นาแห้ว)": "ภูสวนทราย",
    "ต้นสักใหญ่ (คลองตรอน)": "ต้นสักใหญ่",
    "ดอยสอยมาลัย": "ดอยสอยมาลัย-ไม้กลายเป็นหิน",
    "ภูผาเทิบ (มุกดาหาร)": "ภูผาเทิบ",
    "ผาแดง (เชียงดาว)": "ผาแดง"
}
```

```python
khao_khitchakut_match = (
    status_name_matches
    .loc[
        status_name_matches["tourism_name"] == "เขาคิชฌกูฎ",
        "closest_status_name"
    ]
    .iloc[0]
)

status_mapping["เขาคิชฌกูฎ"] = khao_khitchakut_match
```

```python
tourism["park_status_final"] = (
    tourism["park_status_key"]
    .replace(status_mapping)
)
```

```python
remaining_status_unmatched = sorted(
    set(tourism["park_status_final"]) -
    set(status["park_status_key"])
)

print("จำนวนที่ยังจับคู่ไม่ได้:")
print(len(remaining_status_unmatched))

print("\nรายชื่อ:")
print(remaining_status_unmatched)
```

13. ตรวจชื่อปัจจุบันของอุทยานที่ยังจับคู่ไม่ได้

ค้นหาชื่ออุทยานที่เกี่ยวข้องกับชื่อเก่าหรือชื่อที่เปลี่ยนแปลง
ในชุดข้อมูลเปิด-ปิด ก่อนกำหนด Mapping ขั้นสุดท้าย

```python
search_terms = [
    "ภูสระ",
    "ภูหินจอมธาตุ",
    "ภูพระบาท",
    "แควน้อย",
    "แม่เงา",
    "แม่ยวม"
]

for term in search_terms:
    matches = sorted(
        status.loc[
            status["park_status_key"].str.contains(
                term,
                na=False,
                regex=False
            ),
            "park_status_key"
        ].unique()
    )

    print(f"\nค้นหา: {term}")
    print(matches)
```

## 14. กำหนด Mapping ขั้นสุดท้ายสำหรับชื่ออุทยาน

กำหนดชื่อมาตรฐานสำหรับอุทยานที่ตรวจสอบความสัมพันธ์ของชื่อเก่าและชื่อปัจจุบันแล้ว
และไม่บังคับจับคู่กรณีที่ยังมีความไม่แน่นอน

```python
final_status_mapping = {
    "ภูสระดอกบัว": "ภูสระบัว",
    "ภูผายา": "ภูหินจอมธาตุ-ภูพระบาท",
    "แก่งเจ็ดแคว": "แควน้อย"
}

tourism["park_status_final"] = (
    tourism["park_status_final"]
    .replace(final_status_mapping)
)

remaining_status_unmatched = sorted(
    set(tourism["park_status_final"]) -
    set(status["park_status_key"])
)

print("จำนวนที่ยังจับคู่ไม่ได้:")
print(len(remaining_status_unmatched))

print("\nรายชื่อ:")
print(remaining_status_unmatched)
```

15. สร้าง Availability Dataset ขั้นสุดท้าย

สร้างข้อมูลสถานะการเปิด-ปิดรายเดือนสำหรับอุทยานทั้งหมดในชุดข้อมูลนักท่องเที่ยว
โดยเชื่อมกับข้อมูลเปิด-ปิดที่ตรวจสอบชื่อแล้ว และเก็บกรณีที่ไม่มีข้อมูลยืนยัน
เป็น Unknown แทนการสมมติว่าเปิดหรือปิด

```python
import numpy as np

# สร้างสถานะรายเดือนจากข้อมูลเปิด-ปิดใหม่
availability_rows = []

for _, row in status.iterrows():

    for month in range(1, 13):

        availability_rows.append({
            "park_status_key": row["park_status_key"],
            "attraction_name": row["ชื่อแหล่งท่องเที่ยว"],
            "month": month,
            "status": get_month_status(row, month)
        })

status_monthly = pd.DataFrame(availability_rows)


# สรุปจากระดับแหล่งท่องเที่ยว -> ระดับอุทยาน
status_park_month = (
    status_monthly
    .groupby(["park_status_key", "month"])
    .agg(
        total_attractions=("attraction_name", "count"),
        open_attractions=("status", lambda x: (x == "Open").sum()),
        closed_attractions=("status", lambda x: (x == "Closed").sum()),
        unknown_attractions=("status", lambda x: (x == "Unknown").sum())
    )
    .reset_index()
)

status_park_month["known_attractions"] = (
    status_park_month["open_attractions"]
    + status_park_month["closed_attractions"]
)

status_park_month["availability_pct"] = (
    status_park_month["open_attractions"]
    / status_park_month["known_attractions"]
    * 100
)

status_park_month["unknown_pct"] = (
    status_park_month["unknown_attractions"]
    / status_park_month["total_attractions"]
    * 100
)
```

```python
# รายชื่ออุทยาน 165 แห่งของ Tourism
tourism_park_map = (
    tourism[
        ["park_name", "park_status_final"]
    ]
    .drop_duplicates()
)

# เดือน 1-12
months = pd.DataFrame({
    "month": range(1, 13)
})

# 165 อุทยาน × 12 เดือน
park_month_base = tourism_park_map.merge(
    months,
    how="cross"
)

# เชื่อมกับ Availability
park_availability_final = park_month_base.merge(
    status_park_month,
    left_on=["park_status_final", "month"],
    right_on=["park_status_key", "month"],
    how="left"
)

# ระบุว่ามีข้อมูล Availability หรือไม่
park_availability_final["availability_data_status"] = np.where(
    park_availability_final["availability_pct"].notna(),
    "Known",
    "Unknown"
)
```
